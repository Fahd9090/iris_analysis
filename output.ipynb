```python
import pandas as pd

df = pd.read_csv('data/iris.csv')
species_summary = df.groupby('species').mean()
species_summary.to_csv('output/species_summary.csv')
```

### In this cell we grouped the data by species calculated the mean values and saved the summary dataframe to a csv file

```python
filtered_df = df[df['sepal_length'] > 6]
filtered_df.to_csv('output/filtered_iris.csv', index=False)
```

### In this cell we filtered the dataset for rows where sepal length is greater than 6 and saved the filtered dataframe to a csv file

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df, 
    x='petal_length', 
    y='petal_width', 
    hue='species', 
    style='species', 
    s=100
)
```

### In this cell we detailed the specific characteristics of each species showing that setosa has the smallest petals versicolor has medium measurements and virginica has the largest petal length and width

```python


plt.figure(figsize=(10, 6))

sns.boxplot(x='species', y='petal_length', data=df, palette='Set2')

plt.title(
    'مقارنة توزيع طول البتلة (Petal Length) لأنواع الآريس',
    fontsize=14,
    pad=15,
)
plt.xlabel('نوع الزهرة (Species)', fontsize=12)
plt.ylabel('طول البتلة (Petal Length)', fontsize=12)

plt.show()
```

### In this cell we created a box plot comparing petal length distributions across species revealing distinct median values and variability with setosa showing the lowest values versicolor in the middle and virginica having the highest length values
